# 🎤 Risk Audio Analysis - Streamlit App

This notebook will start the Risk Audio Analysis Streamlit app in JupyterLab.

## Steps:
1. Run the cell below
2. Look for 'Port Forwarding' in the left sidebar
3. Add port 8080
4. Click 'Open in Browser'

## Alternative:
- Try: http://localhost:8080


```python
import subprocess
import sys
import os
import time
from IPython.display import display, HTML

# Set environment variables for JupyterLab
os.environ['STREAMLIT_SERVER_PORT'] = '8080'
os.environ['STREAMLIT_SERVER_ADDRESS'] = '0.0.0.0'
os.environ['STREAMLIT_SERVER_HEADLESS'] = 'true'
os.environ['STREAMLIT_BROWSER_GATHER_USAGE_STATS'] = 'false'
os.environ['STREAMLIT_SERVER_ENABLE_CORS'] = 'false'

print("🚀 Starting Risk Audio Analysis Streamlit app...")
print("📱 Access instructions:")
print("   1. Look for 'Port Forwarding' in JupyterLab left sidebar")
print("   2. Add port 8080")
print("   3. Click 'Open in Browser'")
print("   4. Or try: http://localhost:8080")
print("\n🛑 To stop: Interrupt this cell (Ctrl+C)")

# Start Streamlit
try:
    subprocess.run([
        sys.executable, '-m', 'streamlit', 'run', 'streamlit_app.py',
        '--server.port', '8080',
        '--server.address', '0.0.0.0',
        '--server.headless', 'true',
        '--browser.gatherUsageStats', 'false',
        '--server.enableCORS', 'false',
        '--server.enableXsrfProtection', 'false'
    ])
except KeyboardInterrupt:
    print("\n👋 Streamlit server stopped")
```
